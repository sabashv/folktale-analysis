```python
import pandas as pd
import re
import datetime
```

```python
# Direct link to the raw CSV
url = "https://raw.githubusercontent.com/j-hagedorn/trilogy/master/data/aft.csv"
df_stories = pd.read_csv(url)
print(df_stories[['tale_title', 'atu_id', 'text']].head())
```

```python
df_stories['source'][0]
```

```python
def extract_year(source_string):
    if isinstance(source_string, str):
        # four-digit number anywhere in text inside parentheses
        match = re.search(r'\(.*(\d{4}).*\)', source_string)
        if match:
            year = int(match.group(1))
            if 1000 <= year <= 2025:
                return year
    return None

df_stories['publication_year'] = df_stories['source'].apply(extract_year)

# Convert to nullable integer type to remove .0 and handle NaN values
df_stories['publication_year'] = df_stories['publication_year'].astype(pd.Int64Dtype())

print(df_stories[['tale_title', 'source', 'publication_year']].head())
```

```python
df_stories['publication_year']
```

```python
missing_years_count = df_stories['publication_year'].isna().sum()
print(f"Number of rows with missing publication year: {missing_years_count}")
```

```python
df_stories.to_csv('aft_with_year.csv', index=False)
```
